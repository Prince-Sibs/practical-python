## Exercise 1

- 1. Write a program that asks the user to enter a string. The program should then print the
following:<br><br>
(a) The total number of characters in the string<br><br>
(b) The string repeated 10 times<br><br>
(c) The first character of the string (remember that string indices start at 0)<br><br>
(d) The first three characters of the string<br><br>
(e) The last three characters of the string<br><br>
(f) The string backwards<br><br>
(g) The seventh character of the string if the string is long enough and a message otherwise<br><br>
(h) The string with its first and last characters removed<br><br>
(i) The string in all caps<br><br>
(j) The string with every a replaced with an e<br><br>
(k) The string with every letter replaced by a space<br><br>

```python
# Enter a sting
s = input("Enter a string: ")

# Print len of string
print("Chars of string are", len(s))

# Print s*10
print(s*10)

# Print 1st char in s
print(s[0])

# Print 1st 3 chars in s
print(s[:3])

# Print last 3 char in s
print(s[-3:])

# Print s in backward
print(s[::-1])

if len(s) >= 7:
    print(s[6])
else:
    print("The string has no >=7 chars")

# Print s with no its 1st and last char
print(s[1:len(s) - 1])

# Print s in caps
print(s.upper())

# Print replacing a with e
print(s.replace('a', 'e'))

# Print replacing every letter with a space
print(len(s)*' ')
```

## Exercise 3

-  People often forget closing parentheses when entering formulas. Write a program that asks
the user to enter a formula and prints out whether the formula has the same number of opening and closing parentheses.

```python
formula = input("Enter a formula: ")

op = formula.count('(')
cp = formula.count(')')

if op == cp:
    print("same number of parentheses")
else:
    print("not same number of parentheses")
```

## Exercise 7

- Write a program that asks the user to enter a word and determines whether the word is a
palindrome or not. A palindrome is a word that reads the same backwards as forwards

```python
w = input('Enter a word: ')

if w == w[::-1]:
    print('Word is a palindrome')
else:
    print("Word is not a palindrome")
```

## Exercise 8

- At a certain school, student email addresses end with @student.college.edu, while professor email addresses end with @prof.college.edu. Write a program that first asks the
user how many email addresses they will be entering, and then has the user enter those addresses. After all the email addresses are entered, the program should print out a message
indicating either that all the addresses are student addresses or that there were some professor addresses entered.

```python
f = int(input("Enter how many emails: "))

flag = 0
for i in range(f):
    email = input("Enter email: ")

    if '@prof.college.edu' in email:
        flag = 1

if flag == 1:
    print('Some professor addresses were entered')
else:
    print("All addresses were student addresses")
```
