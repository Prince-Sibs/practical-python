## Exercise 1

-  Write a program that counts how many of the squares of the numbers from 1 to 100 end in a
1

```python
count = 0
for i in range(1, 101):
    if (i**2) % 10 == 1:
        count = count + 1

print(count, 'squares end in a 1')
```

## Exercise 2

-  Write a program that counts how many of the squares of the numbers from 1 to 100 end in a
4 and how many end in a 9

```python
count_1 = 0
count_2 = 0
for i in range(1, 101):
    if (i**2) % 10 == 4:
        count_1 = count_1 + 1
    elif (i**2) % 10 == 9:
        count_2 = count_2 + 1

print(count_1, 'sqaures end in a 4')
print(count_2, 'sqaures end in a 9')
```

-  Write a program to compute the sum 1−2+3−4+···+1999−2000

```python
total = 0
for i in range(1, 2001):
    if i%2 ==1:
        total = total + i
    else:
        total = total - i
print(total)
```
